```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "Age")
plt.xlabel("Attrition")
plt.ylabel("Age")
plt.title("Attrition Vs Age")
plt.show()
```

### findings: Employees with attrition tend to be younger on average, suggesting early-career employees may have a higher tendency to switch jobs in search for better opportunities, or career exploration

```python
# WorkLifeBalance
# 1 'Bad'
# 2 'Good'
# 3 'Better'
# 4 'Best'

plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "WorkLifeBalance")
plt.xlabel("Attrition")
plt.ylabel("Work Life Balance")
plt.title("Attrition Vs Work Life Balance")
plt.show()
```

### Findings: The difference in work-life balance scores between employees who left and stayed is marginal and inconclusive, other factors likely play a stronger role in attrition.

```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Department", y = "DistanceFromHome", hue = "Attrition")
plt.xlabel("Department")
plt.ylabel("Distance From Home")
plt.title("Attrition in Department Based on Distance from Home")
plt.show()
```

### Findings: Employees with greater commuting distance, espeacially in the HR department, show relatively higher attrition, suggestion travel burden may contribute to employee turnover.

```python
# Income-to-education ratio

# Education
# 1 'Below College'
# 2 'College'
# 3 'Bachelor'
# 4 'Master'
# 5 'Doctor'

plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Education", y = "MonthlyIncome", hue = "Attrition")
plt.xlabel("Education")
plt.ylabel("Monthly Income")
plt.title("Average Monthly Income by Education Level and Attrition")
plt.show()
```

### Findings: Across all education levels, employees who left consistently earned less than those who stayed, suggesting compensation gaps, not education level itself, may be a stronger driver of attrition.

```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "YearsSinceLastPromotion")
plt.xlabel("Attrition")
plt.ylabel("Years Since Last Promotion")
plt.title("Attrition based on Year since last Promotion")
plt.show()
```

### Findings: Employees who left had been promoted more recently on average, suggesting promotion alone does not prevent attrition, and factors like compensation or career growth opportunities likely matter more.
