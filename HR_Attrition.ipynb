```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "Age")
plt.xlabel("Attrition")
plt.ylabel("Age")
plt.title("Attrition Vs Age")
plt.show()
```

Employees with attrition tend to be younger on average, suggesting early-career employees may have a higher tendency to switch jobs in search for better opportunities, or career exploration

```python
# WorkLifeBalance
# 1 'Bad'
# 2 'Good'
# 3 'Better'
# 4 'Best'

plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "WorkLifeBalance")
plt.xlabel("Attrition")
plt.ylabel("Work Life Balance")
plt.title("Attrition Vs Work Life Balance")
plt.show()
```

Employees with lower work-life balance scores show slightly higher attrition, indicating work-life satisfaction my influence employee retention.

```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Department", y = "DistanceFromHome", hue = "Attrition")
plt.xlabel("Department")
plt.ylabel("Distance From Home")
plt.title("Attrition in Department Based on Distance from Home")
plt.show()
```

Employees with greater commuting distance, espeacially in the HR department, show relatively higher attrition, suggestion travel burden may contribute to employee turnover.

```python
# Income-to-education ratio

# Education
# 1 'Below College'
# 2 'College'
# 3 'Bachelor'
# 4 'Master'
# 5 'Doctor'

plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Education", y = "MonthlyIncome", hue = "Attrition")
plt.xlabel("Monthly Income")
plt.ylabel("Education")
plt.title("Average Monthly Income by Education Level and Attrition")
plt.show()
```

Attrition is observed across all education levels, though employeers with higher education and income may still leave, possibily due to expectations for better compenssation or career growth.

```python
plt.figure(figsize = (10,6))
sns.barplot(data = df, x = "Attrition", y = "YearsSinceLastPromotion")
plt.xlabel("Attrition")
plt.ylabel("Years Since Last Promotion")
plt.title("Attrition based on Year since last Promotion")
plt.show()
```

Employees who left tend to have fewer years since their last promotion, suggesting recent promotions alone may not be sufficient to improve long-term retention.
